```python
import csv

pitches = []

def load_pitch_data():
    try:
        
        with open("pitches.txt", "r") as file:
            reader = csv.DictReader(file)

        
            for row in reader:
                pitch = {
                    "site_location": row["Site Location"].strip(),
                    "site_name": row["Site Name"].strip(),
                    "has_pool": row["Has Pool"].strip(),
                    "pitch_number": row["Pitch Number"].strip(),
                    "pitch_length": row["Pitch Length"].strip(),
                    "pitch_width": row["Pitch Width"].strip(),
                    "minimum_stay": row["Minimum Stay"].strip(),
                    "maximum_stay": row["Maximum Stay"].strip(),
                    "pitch_id": row["Pitch ID"].strip()
                }

        
                pitches.append(pitch)

        print("Pitch data loaded successfully.")
        print("Number of pitches loaded:", len(pitches))

    except FileNotFoundError:
        print("Error: pitches.txt could not be found.")

    except KeyError:
        print("Error: The column headings in pitches.txt do not match.")

    except Exception as error:
        print("An error occurred:", error)



load_pitch_data()

for pitch in pitches:
    print(pitch)
```
